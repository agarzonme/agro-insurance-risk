```python
import duckdb
con = duckdb.connect("../data/processed/agro.duckdb", read_only=True)
```

###Periodos

```python
con.sql("""SELECT DISTINCT period
FROM yields
ORDER BY period;
""").df()
```

###Rendimiento promedio por región

```python
con.sql("""SELECT region,
       ROUND(AVG(yield_t_ha), 2) AS avg_yield
FROM yields
WHERE period >= '2008-I'
GROUP BY region
ORDER BY avg_yield DESC;""").df()
```

La serie de Casanare

```python
con.sql("""SELECT period,
       ROUND(yield_t_ha, 2) AS yield_t_ha
FROM yields
WHERE region = 'Casanare'
  AND period >= '2008-I'
ORDER BY period;""").df()
```

###Los peores y los mejores semestres

```python
con.sql("""select period, round(yield_t_ha, 2) as yield_t_ha from yields where region = 'Casanare' and period >= '2008-I' order by yield_t_ha desc;""").df() 
```
